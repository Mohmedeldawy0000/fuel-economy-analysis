Step 6: Fix Data Types

Step 7: Split Combined Fuel Rows

```python
df_08_combined = df_08[df_08['fuel'].str.contains('/')]
df_18_combined = df_18[df_18['fuel'].str.contains('/')]
```

```python
df_08 = df_08[~df_08['fuel'].str.contains('/')]
df_18 = df_18[~df_18['fuel'].str.contains('/')]
```

```python
df1_08 = df_08_combined.copy()
df2_08 = df_08_combined.copy()

df1_18 = df_18_combined.copy()
df2_18 = df_18_combined.copy()
```

```python
split_columns = [
    'fuel',
    'air_pollution_score',
    'city_mpg',
    'hwy_mpg',
    'cmb_mpg',
    'greenhouse_gas_score'
]
```

```python
for col in split_columns:
    df1_08[col] = df1_08[col].astype(str)
    df2_08[col] = df2_08[col].astype(str)

    df1_18[col] = df1_18[col].astype(str)
    df2_18[col] = df2_18[col].astype(str)
```

Question 1
Did vehicle fuel efficiency improve from 2008 to 2018?

```python
# Average Combined MPG
avg_08 = df_08['cmb_mpg'].mean()
avg_18 = df_18['cmb_mpg'].mean()

print("Average Combined MPG (2008):", avg_08)
print("Average Combined MPG (2018):", avg_18)
```

```python
plt.figure(figsize=(6,4))

plt.bar(['2008', '2018'], [avg_08, avg_18])

plt.title('Average Combined MPG')
plt.xlabel('Year')
plt.ylabel('Combined MPG')

plt.show()
```

Question 2
Which fuel type has the highest average fuel economy?

```python
fuel_avg = df_18.groupby('fuel')['cmb_mpg'].mean().sort_values(ascending=False)

print(fuel_avg)
```

```python
plt.figure(figsize=(8,5))

fuel_avg.plot(kind='bar')

plt.title('Average Combined MPG by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Combined MPG')

plt.show()
```

Question 3
Does engine displacement affect fuel economy?

```python
plt.figure(figsize=(7,5))

plt.scatter(df_18['displ'], df_18['cmb_mpg'])

plt.title('Engine Displacement vs Combined MPG')
plt.xlabel('Engine Displacement')
plt.ylabel('Combined MPG')

plt.show()
```

Question 4
Which vehicle class has the best fuel economy?

```python
vehicle_avg = df_18.groupby('veh_class')['cmb_mpg'].mean().sort_values(ascending=False)

print(vehicle_avg)
```

```python
plt.figure(figsize=(10,6))

vehicle_avg.plot(kind='bar')

plt.title('Average Combined MPG by Vehicle Class')
plt.xlabel('Vehicle Class')
plt.ylabel('Average Combined MPG')

plt.xticks(rotation=45)

plt.show()
```

Question 5
Is there a relationship between greenhouse gas score and fuel economy?

```python
plt.figure(figsize=(7,5))

plt.scatter(df_18['greenhouse_gas_score'], df_18['cmb_mpg'])

plt.title('Greenhouse Gas Score vs Combined MPG')
plt.xlabel('Greenhouse Gas Score')
plt.ylabel('Combined MPG')

plt.show()
```
